```sql
USE CATALOG workspace;
USE SCHEMA bde;

-- Minimal view just for Q1
CREATE OR REPLACE TEMP VIEW q1_base AS
SELECT
  date_format(pickup_datetime,'yyyy-MM') AS pickup_month,
  date_format(pickup_datetime,'EEEE')    AS pickup_dow,
  hour(pickup_datetime)                  AS pickup_hour,
  passenger_count,
  total_amount * 0.93                    AS total_amount_eur   
FROM workspace.bde.taxi_trips_final;


```

Q1 a. What was the total number of trips?

```sql
SELECT pickup_month, COUNT(*) AS total_trips
FROM q1_base
GROUP BY pickup_month
ORDER BY pickup_month;

```

Q1 b. Which day of week (e.g. monday, tuesday, etc..) had the most trips?

```sql
SELECT pickup_month, pickup_dow, trips
FROM (
  SELECT pickup_month, pickup_dow, COUNT(*) AS trips,
         ROW_NUMBER() OVER (PARTITION BY pickup_month ORDER BY COUNT(*) DESC) AS r
  FROM q1_base
  GROUP BY pickup_month, pickup_dow
)
WHERE r = 1
ORDER BY pickup_month;
```

Q1 c. Which hour of the day had the most trips?

```sql
SELECT pickup_month, pickup_hour, trips
FROM (
  SELECT pickup_month, pickup_hour, COUNT(*) AS trips,
         ROW_NUMBER() OVER (PARTITION BY pickup_month ORDER BY COUNT(*) DESC) AS r
  FROM q1_base
  GROUP BY pickup_month, pickup_hour
)
WHERE r = 1
ORDER BY pickup_month;
```

Q1 d. What was the average number of passengers?

```sql
SELECT pickup_month, ROUND(AVG(passenger_count), 2) AS avg_passengers
FROM q1_base
GROUP BY pickup_month
ORDER BY pickup_month;
```

Q1 e. What was the average amount paid per trip?

```sql
SELECT
  date_format(pickup_datetime,'yyyy-MM') AS pickup_month,
  ROUND(AVG(total_amount), 2)            AS avg_usd_per_trip
FROM workspace.bde.taxi_trips_final
GROUP BY pickup_month
ORDER BY pickup_month;
```

Q1 f. What was the average amount in paid per passenger (using total_amount)?

Q2. For each taxi colour (yellow and green):

```sql
CREATE OR REPLACE TEMP VIEW q2_base AS
SELECT
  service_type,
  /* minutes, km, km/h */
  (unix_timestamp(dropoff_datetime)-unix_timestamp(pickup_datetime))/60.0 AS duration_min,
  trip_distance*1.60934                                                    AS distance_km,
  CASE WHEN (unix_timestamp(dropoff_datetime)-unix_timestamp(pickup_datetime))>0
       THEN (trip_distance*1.60934) / ((unix_timestamp(dropoff_datetime)-unix_timestamp(pickup_datetime))/3600.0)
  END                                                                      AS speed_kph
FROM workspace.bde.taxi_trips_final;
```

Q2 a. What was the average, median, minimum and maximum trip duration in minutes (with 2 decimals, eg. 90 seconds = 1.50 min)?

```sql
SELECT
  service_type,                                                -- 'yellow' / 'green'
  ROUND(AVG( (unix_timestamp(dropoff_datetime)
            - unix_timestamp(pickup_datetime)) / 60.0), 2)     AS dur_avg_min,
  ROUND(percentile_approx( (unix_timestamp(dropoff_datetime)
                           - unix_timestamp(pickup_datetime)) / 60.0, 0.5, 10000), 2)
                                                              AS dur_median_min,
  ROUND(MIN( (unix_timestamp(dropoff_datetime)
            - unix_timestamp(pickup_datetime)) / 60.0), 2)     AS dur_min_min,
  ROUND(MAX( (unix_timestamp(dropoff_datetime)
            - unix_timestamp(pickup_datetime)) / 60.0), 2)     AS dur_max_min
FROM workspace.bde.taxi_trips_final
GROUP BY service_type
ORDER BY service_type;
```

Q2 b. What was the average, median, minimum and maximum trip distance in km?

```sql
SELECT
  service_type,
  ROUND(AVG(distance_km),2)                                AS dist_avg_km,
  ROUND(percentile_approx(distance_km,0.5,10000),2)        AS dist_med_km,
  ROUND(MIN(distance_km),2)                                AS dist_min_km,
  ROUND(MAX(distance_km),2)                                AS dist_max_km
FROM q2_base
GROUP BY service_type
ORDER BY service_type;
```

Q2 c. What was the average, median, minimum and maximum speed in km per hour?

```sql
SELECT
  service_type,
  ROUND(AVG(speed_kph),2)                                  AS speed_avg_kph,
  ROUND(percentile_approx(speed_kph,0.5,10000),2)          AS speed_med_kph,
  ROUND(MIN(speed_kph),2)                                  AS speed_min_kph,
  ROUND(MAX(speed_kph),2)                                  AS speed_max_kph
FROM q2_base
GROUP BY service_type
ORDER BY service_type;
```

Q4. For 2024, compute the share of total revenue contributed by the Top 10 pickup→dropoff borough pairs (ranked by total_amount).

```sql
WITH b AS (
  SELECT pickup_borough, dropoff_borough, total_amount
  FROM workspace.bde.taxi_trips_final
  WHERE year(pickup_datetime)=2024
),
pair_rev AS (
  SELECT pickup_borough, dropoff_borough, SUM(total_amount) AS revenue_2024
  FROM b
  GROUP BY pickup_borough, dropoff_borough
),
top10 AS (
  SELECT pickup_borough, dropoff_borough, revenue_2024,
         SUM(revenue_2024) OVER () AS total_rev_2024
  FROM pair_rev
  ORDER BY revenue_2024 DESC
  LIMIT 10
)
SELECT
  pickup_borough, dropoff_borough,
  ROUND(revenue_2024,2)                         AS revenue_2024_usd,
  ROUND(revenue_2024/total_rev_2024,4)          AS share_of_total_2024
FROM top10
ORDER BY revenue_2024_usd DESC;
```

Q7. Classify each trip into bins of durations:
 a. Under 50 Mins
 b. From 50 mins to 100 mins
 c. From 10 mins to 200 mins
 d. From 200 mins to 300 mins
 e. From 300 mins to 600 mins
 f. At least 600 mins *10

```sql
CREATE OR REPLACE TEMP VIEW q7_base AS
SELECT
  CASE
    WHEN (unix_timestamp(dropoff_datetime)-unix_timestamp(pickup_datetime))/60.0 < 5   THEN 'Under 5 Mins'
    WHEN (unix_timestamp(dropoff_datetime)-unix_timestamp(pickup_datetime))/60.0 < 10  THEN '5–10 Mins'
    WHEN (unix_timestamp(dropoff_datetime)-unix_timestamp(pickup_datetime))/60.0 < 20  THEN '10–20 Mins'
    WHEN (unix_timestamp(dropoff_datetime)-unix_timestamp(pickup_datetime))/60.0 < 30  THEN '20–30 Mins'
    WHEN (unix_timestamp(dropoff_datetime)-unix_timestamp(pickup_datetime))/60.0 < 60  THEN '30–60 Mins'
    ELSE 'At least 60 Mins'
  END AS duration_bin,
  trip_distance*1.60934 AS distance_km,
  CASE WHEN (unix_timestamp(dropoff_datetime)-unix_timestamp(pickup_datetime))>0
       THEN (trip_distance*1.60934) / ((unix_timestamp(dropoff_datetime)-unix_timestamp(pickup_datetime))/3600.0)
  END AS speed_kph,
  total_amount
FROM workspace.bde.taxi_trips_final;
```

Then for each bins, calculate: 
a. Average speed /10(km per hour)
b. Average distance per aud dollar /10(km per $)

```sql
SELECT
  duration_bin,
  ROUND(AVG(speed_kph),2) AS avg_speed_kph,
  ROUND(AVG(CASE WHEN total_amount>0 THEN distance_km/total_amount END),4) AS avg_km_per_dollar
FROM q7_base
GROUP BY duration_bin
ORDER BY
  CASE duration_bin
    WHEN 'Under 5 Mins' THEN 1
    WHEN '5–10 Mins'    THEN 2
    WHEN '10–20 Mins'   THEN 3
    WHEN '20–30 Mins'   THEN 4
    WHEN '30–60 Mins'   THEN 5
    ELSE 6
  END;
```

Q8. Which duration bin will you advise a taxi driver to target to maximise his income?

```sql
WITH b AS (
  SELECT
    CASE
      WHEN (unix_timestamp(dropoff_datetime)-unix_timestamp(pickup_datetime))/60.0 < 5   THEN 'Under 5 Mins'
      WHEN (unix_timestamp(dropoff_datetime)-unix_timestamp(pickup_datetime))/60.0 < 10  THEN '5–10 Mins'
      WHEN (unix_timestamp(dropoff_datetime)-unix_timestamp(pickup_datetime))/60.0 < 20  THEN '10–20 Mins'
      WHEN (unix_timestamp(dropoff_datetime)-unix_timestamp(pickup_datetime))/60.0 < 30  THEN '20–30 Mins'
      WHEN (unix_timestamp(dropoff_datetime)-unix_timestamp(pickup_datetime))/60.0 < 60  THEN '30–60 Mins'
      ELSE 'At least 60 Mins'
    END AS duration_bin,
    total_amount,
    (unix_timestamp(dropoff_datetime)-unix_timestamp(pickup_datetime))/60.0 AS duration_min
  FROM workspace.bde.taxi_trips_final
)
SELECT duration_bin,
       ROUND(AVG(CASE WHEN duration_min>0 THEN total_amount/(duration_min/60.0) END),2) AS avg_revenue_per_hour_usd
FROM b
GROUP BY duration_bin
ORDER BY avg_revenue_per_hour_usd DESC
LIMIT 1;
```
